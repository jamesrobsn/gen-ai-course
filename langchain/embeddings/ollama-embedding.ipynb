```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = OllamaEmbeddings(model="llama3")
```

```python
embeddings
```

```python
r1 = embeddings.embed_documents(["Alpha is the first letter of the Greed alphabet", "Beta is the second letter of the Greed alphabet"])
```

```python
r1[0]
```

```python
len(r1[0])
```

```python
embeddings.embed_query("What is the second letter of the Greed alphabet")
```
