```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain.text_splitter import CharacterTextSplitter

loader = TextLoader("speech.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=30)
docs = text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings = OllamaEmbeddings(model="llama3", base_url="http://localhost:11434")
db = FAISS.from_documents(docs, embeddings)
db
```

```python
query = "What does the speaker believe is the main reason the United States should enter the war?"
```

```python
query = "How does the speaker describe the desired outcome of the war?"
db_docs = db.similarity_search(query)
db_docs[0].page_content
```

```python
retriever = db.as_retriever(search_type="similarity", search_kwargs={"k": 1})
retriever.invoke(query)
db_docs[0].page_content
```

```python
docs_and_score = db.similarity_search_with_score(query)
```

```python
docs_and_score
```

```python
embedding_vector = embeddings.embed_query(query)
embedding_vector
```

```python
docs_score = db.similarity_search_by_vector(embedding_vector)
docs_score
```

```python
## Saving and loading
db.save_local("faiss_index")
```

```python
new_db = FAISS.load_local("faiss_index", embeddings, allow_dangerous_deserialization=True)
docs = new_db.similarity_search(query)
```

This Jupyter Notebook demonstrates how to use LangChain with FAISS for semantic search over a text document. Here’s a step-by-step explanation:

### 1. **Document Loading and Splitting**
- Loads a text file (`speech.txt`) using `TextLoader`.
- Splits the loaded document into chunks (size 1000, overlap 30) for better embedding and retrieval.

### 2. **Embeddings and Vector Store**
- Uses `OllamaEmbeddings` (with the `llama3` model running locally) to generate embeddings for each chunk.
- Stores these embeddings in a FAISS vector database (`db`).

### 3. **Semantic Search**
- Performs similarity search using a query (e.g., "How does the speaker describe the desired outcome of the war?").
- Retrieves the most relevant document chunk(s) based on vector similarity.

### 4. **Retriever Interface**
- Converts the FAISS DB into a retriever object for more flexible querying.
- Uses `.invoke()` to get the top result.

### 5. **Scored Search**
- Uses `similarity_search_with_score` to get both the document chunk and its similarity score.

### 6. **Direct Vector Search**
- Embeds a query directly and searches by its vector representation.

### 7. **Persistence**
- Saves the FAISS index to disk (`faiss_index`).
- Loads it back and performs search again to demonstrate persistence.

---

**Summary:**  
This notebook walks through loading a document, chunking it, embedding with Ollama, storing in FAISS, performing semantic search, and saving/loading the index. It’s a practical workflow for building a local semantic search engine using LangChain and FAISS.
